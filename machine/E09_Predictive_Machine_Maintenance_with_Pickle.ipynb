### Predictive Machine Maintenance using Markov Process

#### 1.Title

Machine Failure Prediction using Markov Chain Model

#### 2. Objective

To **predict machine health states and failure probability** using a **Markov Process**, so that preventive maintenance can be scheduled to reduce downtime.

#### 3. Problem Statement

In manufacturing industries, machines gradually **degrade over time**. If maintenance is not performed at the right time, machines may fail and stop production.

This project uses a **Markov chain model** to simulate how a machine transitions between different health states and predicts when maintenance is required.

#### 4. Machine States

Define machine condition as states.

| State | Meaning           |
| ----- | ----------------- |
| S1    | Fully Operational |
| S2    | Minor Wear        |
| S3    | Major Fault       |
| S4    | Machine Failure   |

#### 5. Transition Probability Matrix

Example probability matrix:

| Current State | S1  | S2  | S3  | S4  |
| ------------- | --- | --- | --- | --- |
| S1            | 0.7 | 0.2 | 0.1 | 0   |
| S2            | 0.2 | 0.5 | 0.2 | 0.1 |
| S3            | 0   | 0.3 | 0.4 | 0.3 |
| S4            | 0   | 0   | 0   | 1   |

**Interpretation:**
- Healthy machines may move to minor wear
- Minor wear may develop into major faults
- Major faults may lead to failure

#### 6. Implementation Steps

##### Step 1: Define Machine States

```python
states = ["Operational","Minor Wear","Major Fault","Failure"]
```

##### Step 2: Define Transition Matrix

```python
import numpy as np

P = np.array([
[0.7,0.2,0.1,0.0],
[0.2,0.5,0.2,0.1],
[0.0,0.3,0.4,0.3],
[0.0,0.0,0.0,1.0]
])
```

##### Step 3: Simulate Machine Behavior

```python
state = 0

for i in range(15):
    state = np.random.choice([0,1,2,3], p=P[state])
    print("Machine State:", states[state])
```

This shows machine degradation over time.

#### 7. Visualization

```python
import matplotlib.pyplot as plt

history = []
state = 0

for i in range(20):
    history.append(state)
    state = np.random.choice([0,1,2,3], p=P[state])

plt.plot(history)
plt.title("Machine State Transition Over Time")
plt.xlabel("Time")
plt.ylabel("State")
plt.show()
```

This implementation shows how Markov Processes can simulate machine health transitions and support predictive maintenance decisions in manufacturing systems.

### Conclusion

In this experiment, a Markov process was used to model manufacturing processes and machine maintenance. The system states (such as working, degraded, and failed) were analyzed with transition probabilities. This helped in predicting future states and planning maintenance effectively. Overall, the Markov model proved useful for improving reliability and optimizing maintenance strategies.

#### Step 4: Save the Markov Process Model using Pickle

The original code above is unchanged. This additional cell saves the existing `states` and `P` transition matrix into `machine_maintenance_markov_model.pkl` so it can be reused later.

```python
import pickle
import numpy as np

# Save the existing Markov model variables without changing the original code.
model_data = {
    "states": states,
    "transition_matrix": P,
    "model_type": "Markov Process",
    "project": "Predictive Machine Maintenance"
}

with open("machine_maintenance_markov_model.pkl", "wb") as model_file:
    pickle.dump(model_data, model_file)

print("Pickle model saved: machine_maintenance_markov_model.pkl")
```

#### Step 5: Load the Pickle Model and Predict the Next Machine State

Run this cell after saving the model. It loads the stored transition matrix and simulates one next-state prediction from the selected current state.

```python
import pickle
import numpy as np

with open("machine_maintenance_markov_model.pkl", "rb") as model_file:
    loaded_model = pickle.load(model_file)

loaded_states = loaded_model["states"]
loaded_P = loaded_model["transition_matrix"]

# Set the current state index: 0=Operational, 1=Minor Wear,
# 2=Major Fault, 3=Failure.
current_state = 0
next_state = np.random.choice(
    range(len(loaded_states)),
    p=loaded_P[current_state]
)

print("Current Machine State:", loaded_states[current_state])
print("Predicted Next Machine State:", loaded_states[next_state])
print("Transition probabilities:", loaded_P[current_state])
```
